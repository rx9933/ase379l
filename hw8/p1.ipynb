```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
def f(sigma, M):
        
    X = np.random.normal(1, sigma, M)
    Y = X**3

    print("Mean:", np.mean(Y))
    print("Variance:", np.var(Y))

    plt.figure()
    plt.hist(Y, bins=200, density=True, alpha=0.5, label="Histogram")
    sns.kdeplot(Y, label="KDE")

    plt.xlabel("y")
    plt.ylabel("Density")
    plt.legend()
    plt.title(f"PDF approximation of Y ($\sigma^2$ = {sigma**2:.3e})")
    plt.show()

M = 10_000_000
sigma2 = 0.01
sigma = np.sqrt(sigma2)
f(sigma, M)
```

```python
M = 10_000_000
sigma2 = 0.1
sigma = np.sqrt(sigma2)
f(sigma, M)
```

# 1.4
The perturbation method provides a good approximation when the input
variance is small. For $\sigma^2 = 0.01$, the Monte Carlo estimates of
the mean and variance are close to the first–order perturbation
predictions, indicating that the linearization of the nonlinear
function $f(X) = X^3$ around the mean is sufficiently accurate.
However, for $\sigma^2 = 0.1$, the empirical mean and variance deviate
significantly from the perturbation estimates. This occurs because the
cubic transformation introduces strong nonlinear effects, and the
first–order Taylor approximation is no longer adequate over the wider
range of possible input values. This illustrates that perturbation
methods are most reliable when the input uncertainty is small or when
the function is approximately linear near the mean. For strongly
nonlinear functions or larger uncertainties, higher–order expansions or
sampling-based methods such as Monte Carlo simulation are required for
accurate uncertainty propagation.
